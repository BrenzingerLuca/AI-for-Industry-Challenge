# 2. Splitting the data into train and test sets


```python
# 1. Define paths (relative to this notebook)
# Adjust the number of '../' depending on your exact folder depth
source_dataset_path = os.path.abspath("../data/datasets/single_nic_card_dataset")
raw_images_dir = os.path.join(source_dataset_path, "images")
raw_labels_dir = os.path.join(source_dataset_path, "labels")

# Target directory for the split data
split_dir = "processed_dataset"

if not os.path.exists(raw_images_dir):
    print(f"Error: Source path not found at: {raw_images_dir}")
else:
    print(f"Source found: {source_dataset_path}")

    # 2. Create folder structure
    for split in ['train', 'val']:
        for sub in ['images', 'labels']:
            os.makedirs(os.path.join(split_dir, split, sub), exist_ok=True)

    # 3. List and split files
    image_files = sorted([f for f in os.listdir(raw_images_dir) if f.endswith('.jpg')])
    names = [os.path.splitext(f)[0] for f in image_files]

    # Split: 80% training, 20% validation
    train_names, val_names = train_test_split(names, test_size=0.2, random_state=42)

    def copy_dataset_files(file_names, target_set):
        for name in file_names:
            shutil.copy(os.path.join(raw_images_dir, f"{name}.jpg"), 
                        os.path.join(split_dir, target_set, "images", f"{name}.jpg"))
            shutil.copy(os.path.join(raw_labels_dir, f"{name}.txt"), 
                        os.path.join(split_dir, target_set, "labels", f"{name}.txt"))

    print("Copying files...")
    copy_dataset_files(train_names, 'train')
    copy_dataset_files(val_names, 'val')

    print(f"Dataset split complete!")
    print(f"Training images: {len(train_names)} | Validation images: {len(val_names)}")
```

# 3. Generate the YOLO Configuration (YAML)

```python
yaml_path = "sfp_pose_config.yaml"
# Get absolute path of the processed dataset
absolute_data_path = os.path.abspath(split_dir)

yaml_content = f"""
path: {absolute_data_path}
train: train/images
val: val/images

# Keypoints: 4 corners (TL, TR, BR, BL)
# Shape: [number of keypoints, number of values (x, y, visibility)]
kpt_shape: [4, 3] 

names:
  0: sfp_port_0
  1: sfp_port_1
"""

with open(yaml_path, "w") as f:
    f.write(yaml_content)

print(f"YAML configuration saved to {yaml_path}")
```
