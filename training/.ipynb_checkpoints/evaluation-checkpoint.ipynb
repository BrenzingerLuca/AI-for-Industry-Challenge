# 1. Setup


```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO
from pathlib import Path

print("Dependencies ready.")
```

# 2. Configuration


```python
MODEL_PATH      = "best.pt"
VAL_IMAGES_PATH = "processed_dataset/val/images"
DATA_YAML       = "sfp_pose_config.yaml"
CONF_THRESHOLD  = 0.25
NUM_IMAGES_TO_SHOW = 12

CLASS_NAMES  = {0: "sfp_port_0", 1: "sfp_port_1"}
CLASS_COLORS = {0: (0, 255, 0), 1: (0, 200, 255)}  # BGR: green, yellow

model = YOLO(MODEL_PATH)
print(f"Model loaded: {MODEL_PATH}")
print(f"Model task:   {model.task}")
```

# 3. Evaluation
Runs the model on the full validation set and computes:
- Precision, Recall
- Box mAP50 / mAP50-95  (detection quality)
- Pose mAP50 / mAP50-95 (keypoint quality — most important for us)

```python
metrics = model.val(data=DATA_YAML, conf=CONF_THRESHOLD)

print("\n" + "="*50)
print("EVALUATION RESULTS")
print("="*50)
print(f"\n{'Metric':<25} {'Value':>10}")
print("-"*37)
print(f"{'Box mAP50':<25} {metrics.box.map50:>10.3f}")
print(f"{'Box mAP50-95':<25} {metrics.box.map:>10.3f}")
print(f"{'Pose mAP50':<25} {metrics.pose.map50:>10.3f}")
print(f"{'Pose mAP50-95':<25} {metrics.pose.map:>10.3f}")
print("="*50)


# ---- Per-class breakdown ----
print(f"\n{'Class':<20} {'Box mAP50':>12} {'Pose mAP50':>12} {'Pose mAP50-95':>14}")
print("-" * 60)
for i, name in CLASS_NAMES.items():
    box_map50  = metrics.box.ap50[i]  if i < len(metrics.box.ap50)  else float('nan')
    pose_map50 = metrics.pose.ap50[i] if i < len(metrics.pose.ap50) else float('nan')
    pose_map   = metrics.pose.ap[i]   if i < len(metrics.pose.ap)   else float('nan')
    print(f"{name:<20} {box_map50:>12.3f} {pose_map50:>12.3f} {pose_map:>14.3f}")


# ---- Bar chart ----
class_labels    = list(CLASS_NAMES.values())
box_map50_vals  = [metrics.box.ap50[i]  for i in CLASS_NAMES]
pose_map50_vals = [metrics.pose.ap50[i] for i in CLASS_NAMES]

x     = np.arange(len(class_labels))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
bars1 = ax.bar(x - width/2, box_map50_vals,  width, label='Box mAP50',  color='steelblue')
bars2 = ax.bar(x + width/2, pose_map50_vals, width, label='Pose mAP50', color='darkorange')

ax.set_ylim(0, 1.05)
ax.set_ylabel('mAP50')
ax.set_title('Box vs Pose mAP50 per Class')
ax.set_xticks(x)
ax.set_xticklabels(class_labels)
ax.legend()
ax.axhline(y=0.9, color='gray', linestyle='--', linewidth=0.8)

for bar in bars1 + bars2:
    ax.annotate(f'{bar.get_height():.3f}',
                xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
                xytext=(0, 4), textcoords='offset points',
                ha='center', fontsize=10)

plt.tight_layout()
plt.savefig('metrics_bar_chart.png', dpi=150)
plt.show()
```

# 4. Visual Inspection
### Keypoint color legend:
- Red    = kp0 Top-Left
- Green  = kp1 Top-Right
- Yellow = kp2 Bottom-Right
- Magenta= kp3 Bottom-Left
- White line kp0->kp1 = top edge, encodes orientation

```python
def draw_predictions(image_bgr, result, conf_threshold=0.25):
    img = image_bgr.copy()
    if result.boxes is None or len(result.boxes) == 0:
        return img

    for i, box in enumerate(result.boxes):
        conf   = float(box.conf)
        if conf < conf_threshold:
            continue
        cls_id = int(box.cls)
        color  = CLASS_COLORS.get(cls_id, (255, 255, 255))
        label  = f"{CLASS_NAMES.get(cls_id, cls_id)} {conf:.2f}"

        x1, y1, x2, y2 = map(int, box.xyxy[0])
        cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
        cv2.putText(img, label, (x1, y1 - 8),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

        if result.keypoints is not None and i < len(result.keypoints):
            kpts = result.keypoints.xy[i].cpu().numpy()
            kpt_colors = [(0,0,255), (0,255,0), (0,255,255), (255,0,255)]
            for k, (kx, ky) in enumerate(kpts):
                cv2.circle(img, (int(kx), int(ky)), 5, kpt_colors[k], -1)
            cv2.line(img,
                     (int(kpts[0][0]), int(kpts[0][1])),
                     (int(kpts[1][0]), int(kpts[1][1])),
                     (255, 255, 255), 2)
    return img


val_images = sorted(Path(VAL_IMAGES_PATH).glob("*.jpg"))[:NUM_IMAGES_TO_SHOW]
print(f"Running inference on {len(val_images)} images...")

results = model.predict(
    source=[str(p) for p in val_images],
    conf=CONF_THRESHOLD,
    verbose=False
)

cols = 3
rows = (len(val_images) + cols - 1) // cols
fig, axes = plt.subplots(rows, cols, figsize=(18, rows * 5))
axes = axes.flatten()

for idx, (img_path, result) in enumerate(zip(val_images, results)):
    img_bgr   = cv2.imread(str(img_path))
    img_drawn = draw_predictions(img_bgr, result, CONF_THRESHOLD)
    img_rgb   = cv2.cvtColor(img_drawn, cv2.COLOR_BGR2RGB)
    n_det     = len(result.boxes) if result.boxes is not None else 0
    axes[idx].imshow(img_rgb)
    axes[idx].set_title(f"{img_path.name}\n{n_det} detection(s)", fontsize=8)
    axes[idx].axis('off')

for j in range(idx + 1, len(axes)):
    axes[j].axis('off')

plt.suptitle('Validation Set — Predicted Bounding Boxes & Keypoints', fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig('predictions_grid.png', dpi=150, bbox_inches='tight')
plt.show()
```

# 5. Confidence Distribution

Shows how confident the model is across all predictions.
A distribution skewed towards high confidence (>0.9) indicates a well-trained model.

```python
all_confs  = []
all_results = model.predict(
    source=VAL_IMAGES_PATH,
    conf=0.01,
    verbose=False
)

for r in all_results:
    if r.boxes is not None:
        all_confs.extend(r.boxes.conf.cpu().numpy().tolist())

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(all_confs, bins=20, color='steelblue', edgecolor='white')
ax.axvline(x=CONF_THRESHOLD, color='red', linestyle='--',
           label=f'Threshold ({CONF_THRESHOLD})')
ax.set_xlabel('Confidence Score')
ax.set_ylabel('Number of Predictions')
ax.set_title('Confidence Score Distribution — Validation Set')
ax.legend()
plt.tight_layout()
plt.savefig('confidence_distribution.png', dpi=150)
plt.show()

print(f"Total predictions: {len(all_confs)}")
print(f"Mean confidence:   {np.mean(all_confs):.3f}")
print(f"Above threshold:   {sum(c >= CONF_THRESHOLD for c in all_confs)} / {len(all_confs)}")
```
